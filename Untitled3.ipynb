```python
df.drop(columns=["Alley", "MasVnrType", "PoolQC", "Fence", "MiscFeature"], inplace=True)
```

```python
df.isnull().sum()[df.isnull().sum() > 0]
```

```python
# Sayısal sütunları ortalama ile doldur
num_cols = df.select_dtypes(include="number").columns
df[num_cols] = df[num_cols].fillna(df[num_cols].mean())

# Yazısal sütunları en sık geçen değerle doldur
cat_cols = df.select_dtypes(include="object").columns
df[cat_cols] = df[cat_cols].fillna(df[cat_cols].mode().iloc[0])

# Kontrol
df.isnull().sum().sum()
```

```python
df.select_dtypes(include="object").columns
```

```python
df = pd.get_dummies(df)
print(df.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
X = df.drop(columns=["SalePrice"])
y = df["SalePrice"]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = RandomForestRegressor(n_estimators=100)
model.fit(X_train, y_train)

tahminler = model.predict(X_test)
print("Ortalama Hata:", mean_absolute_error(y_test, tahminler))
```
